# <span style="color: #E60000;">Notebook 00. Business Understanding</span>

*Tìm hiểu bài toán và nghiệp vụ rủi ro tín dụng*

---

**Mục tiêu:** Hiểu bối cảnh nghiệp vụ cho vay, bài toán Home Credit, vai trò của AI và mục tiêu nghiên cứu trước khi khảo sát dữ liệu.

## I. Giới thiệu dự án

Dự án xây dựng mô hình dự đoán khả năng khách hàng gặp khó khăn khi trả nợ. Mục đích không phải thay chuyên viên tín dụng ra quyết định, mà cung cấp một tín hiệu định lượng để hỗ trợ thẩm định hồ sơ vay.

Bộ dữ liệu Home Credit Default Risk phản ánh bài toán đánh giá rủi ro tín dụng tiêu dùng thông qua thông tin hồ sơ vay hiện tại, lịch sử tín dụng và hành vi thanh toán của khách hàng.


## II. Tìm hiểu nghiệp vụ cho vay

Home Credit cung cấp tín dụng tiêu dùng và cần đánh giá khả năng trả nợ của khách hàng trước khi cấp khoản vay. Quy trình cơ bản gồm: tiếp nhận hồ sơ → thẩm định thông tin → phê duyệt hoặc từ chối → giải ngân → theo dõi việc trả nợ.

Để đánh giá một hồ sơ vay, doanh nghiệp không chỉ dùng thông tin khách hàng khai báo tại thời điểm đăng ký. Các thông tin về lịch sử vay, trả góp, dư nợ và hành vi sử dụng tín dụng cũng giúp tạo nên bức tranh tài chính đầy đủ hơn.

Vì vậy, dữ liệu của dự án được tổ chức quanh hồ sơ vay hiện tại và các nhóm thông tin bổ sung phản ánh lịch sử tín dụng, hành vi thanh toán và việc sử dụng sản phẩm tín dụng.

Hai sai lầm cần cân bằng là cho vay nhầm khách hàng rủi ro cao, làm tăng nợ xấu; và từ chối nhầm khách hàng tốt, làm mất cơ hội kinh doanh. AI giúp cung cấp thêm tín hiệu định lượng để chuyên viên tín dụng xem xét, không thay thế quyết định của con người.

**Bước tiếp theo:** NB01 sẽ xác định cụ thể các nguồn dữ liệu, vai trò của từng bảng và mối quan hệ giữa chúng.


## III. Bài toán Home Credit và biến TARGET

Bài toán là dự đoán khách hàng có khả năng gặp khó khăn khi trả nợ hay không dựa trên hồ sơ vay và lịch sử tín dụng. Đây là bài toán **phân loại nhị phân (Binary Classification)**. 

| Giá trị | Ý nghĩa |
|---|---|
| `TARGET = 0` | Khách hàng không gặp khó khăn trong thanh toán. |
| `TARGET = 1` | Khách hàng gặp khó khăn trong thanh toán. |

Đầu ra mong muốn là xác suất rủi ro để chuyên viên có thể phân nhóm hồ sơ và ưu tiên kiểm tra kỹ các trường hợp cần thiết.


## IV. Mục tiêu nghiên cứu

Nhóm tập trung trả lời các câu hỏi sau:

- Thông tin nào trong hồ sơ vay hiện tại có liên quan đến rủi ro trả nợ?
- Lịch sử tín dụng, khoản vay trước và hành vi thanh toán có giúp nhận diện rủi ro tốt hơn không?
- Có thể tạo các đặc trưng nào phản ánh năng lực tài chính và hành vi tín dụng của khách hàng?
- Mô hình có thể hỗ trợ chuyên viên phân nhóm rủi ro tốt hơn cách đánh giá đơn giản hay không?

Mục tiêu cuối cùng là biến dữ liệu tín dụng nhiều bảng thành thông tin hỗ trợ quyết định và triển khai thành công cụ demo cho người dùng.


## V. Lý do chọn dataset và những thuận lợi – thách thức

Home Credit phù hợp vì dữ liệu có quy mô lớn, nhiều bảng liên kết và bám sát một bài toán nghiệp vụ thực tế. Dataset giúp nhóm rèn luyện kỹ năng đọc dữ liệu, xử lý missing values, tổng hợp bảng phụ và xây dựng mô hình dự báo.

| Góc nhìn | Nội dung chính |
|---|---|
| **Điểm mạnh** | Có `TARGET` rõ ràng và nhiều nguồn dữ liệu tín dụng/hành vi. |
| **Điểm yếu** | Dữ liệu lớn, nhiều missing values, quan hệ 1-n và giá trị bất thường. |
| **Cơ hội** | Có thể hỗ trợ phát hiện sớm hồ sơ rủi ro và phân luồng thẩm định. |
| **Thách thức** | Join hoặc xử lý dữ liệu sai có thể làm mô hình sai lệch nhưng khó phát hiện. |


## VI. Dataset và công nghệ sử dụng

Dataset gồm bảng hồ sơ vay hiện tại `applications` có biến mục tiêu `TARGET` và các bảng lịch sử như `bureau`, `bureau_balance`, `previous_application`, `installments_payments`, `POS_CASH_balance` và `credit_card_balance`. Bảng `applications` sẽ được chia train/test nội bộ ở NB06; các bảng phụ thường có quan hệ một-nhiều, nên cần tổng hợp về cấp khách hàng trước khi đưa vào mô hình.

*Công nghệ sử dụng*:
- PostgreSQL  → lưu trữ, join, aggregation
- Python      → xử lý dữ liệu và Machine Learning
- FastAPI     → cung cấp API dự đoán
- Streamlit   → giao diện demo và dashboard

## VII. Tình huống thực tế, vai trò AI và giá trị hệ thống

Hai khách hàng có thể có cùng thu nhập và cùng số tiền muốn vay, nhưng mức rủi ro khác nhau nếu một người từng thanh toán trễ hoặc có nhiều khoản vay tồn tại. Điều này cho thấy doanh nghiệp cần kết hợp hồ sơ hiện tại với lịch sử tín dụng.

AI học từ các hồ sơ trong quá khứ để tìm tín hiệu rủi ro, sau đó cung cấp xác suất dự đoán cho chuyên viên. Hệ thống có thể giúp tăng tính nhất quán khi thẩm định, rút ngắn thời gian xử lý hồ sơ và hỗ trợ phát hiện nhóm khách hàng cần xem xét kỹ hơn.

Kết quả mô hình luôn là thông tin hỗ trợ; quyết định cho vay vẫn cần được chuyên viên tín dụng kiểm tra trong ngữ cảnh thực tế.
